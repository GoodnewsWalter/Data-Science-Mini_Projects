```python
class ResortRoom:
    def __init__(self, room_number, room_type, nightly_rate):
        self.is_booked = False
        self.room_number = room_number
        self.room_type = room_type
        self.nightly_rate = nightly_rate
        
    
    def book_room(self):
        if not self.is_booked:
            self.is_booked = True
            return f"Room {self.room_number} successfully booked!."
        else:
            return "The room is currently unavailable"
    
    def checkout(self):
        if self.is_booked:
            self.is_booked = False
            return f"Room {self.room_number} successfully checked out!."
        else:
            return "The room is already empty."
    
    def check_status(self):
        if self.is_booked:
            return f"Room {self.room_number} ( {self.room_type} ) - ${self.nightly_rate}/night - Status: Booked"
        
        else:
            return f"Room {self.room_number} ( {self.room_type} ) - ${self.nightly_rate}/night - Status: Available"
```

```python
oasis_villa = ResortRoom(101, "Villa", 850.00)
beachfront_suite = ResortRoom(202, "Suite", 650.00)
garden_room = ResortRoom(303, "Garden Room", 450.00)
```

```python
oasis_villa.book_room()
```

```python
beachfront_suite.book_room()
```

```python
beachfront_suite.checkout()
```

```python
garden_room.check_status()
```
